In this notebook, we load the IT support dataset, inspect its shape, check data types, and review sample records.

```python
import pandas as pd

# Load dataset
df = pd.read_csv('../datasets/synthetic_it_support_tickets.csv')

print("Dataset Shape:", df.shape)
print("\nData Types:")
print(df.dtypes)
print("\nMissing Values:")
print(df.isnull().sum())
```
