# 02 Exploratory Data Analysis (EDA)
In this notebook, we analyze the distribution of target priorities, ticket categories, sentiment, and resolution metrics.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../datasets/synthetic_it_support_tickets.csv')
```

```python
# Set figure style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 5))

# Plot target class distribution
plt.subplot(1, 2, 1)
sns.countplot(data=df, x='priority', order=['low', 'medium', 'high', 'urgent'], palette='Blues_r')
plt.title('Distribution of Priority Levels')

# Plot issue_type by priority
plt.subplot(1, 2, 2)
sns.countplot(data=df, x='issue_type', hue='priority', palette='viridis')
plt.xticks(rotation=45)
plt.title('Issue Type vs Priority')

plt.tight_layout()
plt.show()
```

### Summary of EDA Findings:
* `low` and `medium` priorities make up the majority of support tickets.
* Critical categories like `security_concern` and `bug` tend to have higher proportions of `high` and `urgent` tickets.
