# 04 Feature Engineering
We convert text into numerical TF-IDF vectors and encode categorical metadata variables.

```python
import pandas as pd  # Data manipulation library
import joblib  # For saving fitted feature extractors to disk
from sklearn.feature_extraction.text import TfidfVectorizer  # Text vectorization
from sklearn.preprocessing import OneHotEncoder  # Categorical metadata encoding
```

```python
# Load the cleaned tickets dataset created in the previous step
df = pd.read_csv('../datasets/cleaned_tickets.csv')
```

```python
# Handle missing clean text entries safely
df['cleaned_message'] = df['cleaned_message'].fillna('')
```

```python
# max_features=1500: Expands vocabulary size to capture more domain-specific technical terms
# ngram_range=(1, 2): Captures single words and key 2-word phrases (e.g., 'system crash', 'login error')
# sublinear_tf=True: Applies sublinear scaling (1 + log(tf)) to prevent high-frequency words from dominant weighting
tfidf = TfidfVectorizer(
    max_features=1500,
    ngram_range=(1, 2),
    sublinear_tf=True
)
```

```python
# Learn word vocabulary and transform cleaned ticket messages into numerical sparse feature matrix
X_text = tfidf.fit_transform(df['cleaned_message'])
```

```python
# Specify categorical columns to encode
cat_cols = ['product_area', 'issue_type', 'customer_segment', 'channel', 'platform']
```

```python
# Initialize OneHotEncoder to convert categorical metadata into binary (0/1) indicator columns
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
X_cat = ohe.fit_transform(df[cat_cols])
```

```python
# Export fitted TF-IDF vectorizer and One-Hot Encoder artifacts to the models directory
joblib.dump(tfidf, '../models/tfidf_vectorizer.pkl')
joblib.dump(ohe, '../models/category_encoder.pkl')
```

```python
print("Enhanced Feature Extraction Complete!")
print("TF-IDF Text Matrix Shape:", X_text.shape)
print("Categorical Matrix Shape:", X_cat.shape)
```
