# 03 Data Preprocessing
Here we clean raw text features, handle missing values, and prepare categorical variables.

```python
# Import pandas for data manipulation and working with DataFrames
import pandas as pd

# Import re (regular expressions) for text cleaning and pattern matching (e.g., removing punctuation)
import re

# Import NLTK (Natural Language Toolkit) for text processing tasks
import nltk

# Import stop words to remove common filler words (like "the", "is", "at") that don't add predictive value
from nltk.corpus import stopwords

# Import WordNetLemmatizer to reduce words to their base or dictionary form (e.g., "running" -> "run")
from nltk.stem import WordNetLemmatizer
```

```python
# Download the stop words lexicon from NLTK resources
nltk.download('stopwords')

# Download the WordNet lexical database required for lemmatization
nltk.download('wordnet')
```

```python
# Read the raw CSV file into a Pandas DataFrame from the datasets folder
df = pd.read_csv('../datasets/synthetic_it_support_tickets.csv')
```

```python
# Replace missing (NaN) values in the 'region' column with the string 'Unknown' to avoid null errors
df['region'] = df['region'].fillna('Unknown')
```

```python
# Store English stop words in a set for fast lookup efficiency
stop_words = set(stopwords.words('english'))
```

```python
# Instantiate the WordNet Lemmatizer object
lemmatizer = WordNetLemmatizer()
```

```python
def clean_text(text):
    """
    Cleans raw text by converting to lowercase, removing special characters/numbers,
    filtering out stop words, and applying word lemmatization.
    """
    # Verify if the input is a valid string; return an empty string if it's missing or non-string
    if not isinstance(text, str):
        return ""
    
    # Step A: Convert all text characters to lowercase so words like 'Error' and 'error' are treated identically
    text = text.lower()
    
    # Step B: Remove all punctuation, special characters, and non-alphanumeric characters (keeps only a-z, 0-9, spaces)
    text = re.sub(r'[^a-z0-9\s]', '', text)
    
    # Step C: Split the text sentence string into individual word tokens
    words = text.split()
    
    # Step D: Filter out stop words and apply lemmatization to simplify each remaining word to its root dictionary form
    words = [lemmatizer.lemmatize(word) for word in words if word not in stop_words]
    
    # Step E: Join the processed list of words back into a single cleaned text string separated by spaces
    return " ".join(words)
```

```python
# Run the clean_text function across the raw 'initial_message' column and save results to a new column 'cleaned_message'
df['cleaned_message'] = df['initial_message'].apply(clean_text)
```

```python
# Print header message
print("Sample Cleaned Text:")

# Print side-by-side comparison of original initial_message vs newly created cleaned_message for the first 5 records
print(df[['initial_message', 'cleaned_message']].head())
```

```python
# Export the updated DataFrame containing cleaned text to a new CSV file in the datasets folder (without index numbers)
df.to_csv('../datasets/cleaned_tickets.csv', index=False)
```
